```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from cmath import *
import matplotlib.cm as cm
from mpl_toolkits.mplot3d import Axes3D

sp.init_printing()
```

```python
# 2.1. Представление комплексных выражений в алгебраической, тригонометрической
#      и показательной формах.
print("Задание 2.1")

def complex_forms(expr, name):
    # Упрощаем выражение
    val = sp.simplify(expr)
    # Получаем численное значение для вычисления модуля и аргумента
    num_val = complex(val)
    r = abs(num_val)
    phi = sp.arg(num_val)

    print(f"\n{name}:")
    print(f"Алгебраическая форма: {sp.expand(val)}")
    print(f"Тригонометрическая форма: {r:.5f} * (cos({phi:.5f}) + i*sin({phi:.5f}))")
    print(f"Показательная форма: {r:.5f} * e^(i * {phi:.5f})")
    print(f"Численное значение: {num_val}")

# a) (5+i)(7-6i) / (3+i)
expr_a = (5 + sp.I) * (7 - 6*sp.I) / (3 + sp.I)
complex_forms(expr_a, "a)")

# b) (1+i)^5 / (1-i)^3
expr_b = (1 + sp.I)**5 / (1 - sp.I)**3
complex_forms(expr_b, "b)")

# c) Сложное выражение
term1 = (-1 + sp.I*sp.sqrt(3))**15 / (1 - sp.I)**20
term2 = (-1 - sp.I*sp.sqrt(3))**15 / (1 + sp.I)**20
expr_c = term1 + term2
complex_forms(expr_c, "c)")
```

```python
# 2.2. Решить уравнение z^8 = 1 + i и изобразить на комплексной плоскости
print("Задание 2.2: Решение z^8 = 1 + i")

# Переводим правую часть в показательную форму
r = 2**(1/4)
phi = np.pi / 4  # Arg(1+i) = pi/4
n = 8

roots = []
for k in range(n):
    # Формула корней: z_k = r^(1/n) * e^(i*(phi + 2*pi*k)/n)
    root_r = r**(1/n)
    root_phi = (phi + 2*np.pi*k) / n
    z = root_r * np.exp(1j * root_phi)
    roots.append(z)
    print(f"z_{k} = {z:.5f}")

# Визуализация
plt.figure(figsize=(6, 6))
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.scatter([z.real for z in roots], [z.imag for z in roots], color='red', zorder=5)

# Рисуем окружность
theta = np.linspace(0, 2*np.pi, 100)
plt.plot(roots[0].real * np.cos(theta), roots[0].real * np.sin(theta), 'b--', alpha=0.5)

plt.title("Корни уравнения z^8 = 1 + i")
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")
plt.grid(True)
plt.axis('equal')
plt.show()


```

```python
# 2.3. Вычислить корни и изобразить на комплексной плоскости
print("Задание 2.3: Вычисление корней")

def plot_roots(roots_list, title):
    plt.figure(figsize=(6, 6))
    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)
    plt.scatter([z.real for z in roots_list], [z.imag for z in roots_list], color='green', zorder=5)
    plt.title(title)
    plt.xlabel("Re(z)")
    plt.ylabel("Im(z)")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

# a) sqrt[4]{1}
roots_3a = [np.exp(2j * np.pi * k / 4) for k in range(4)]
print("\n2.3 a) Корни 4-й степени из 1:")
for i, z in enumerate(roots_3a): print(f"z_{i} = {z:.4f}")
plot_roots(roots_3a, "2.3 a) sqrt[4]{1}")

# b) sqrt[4]{ -18 / (1 + i*sqrt(3)) }
val_3b = -18 / (1 + 1j*np.sqrt(3))
r_3b = abs(val_3b)
phi_3b = np.angle(val_3b)
roots_3b = [r_3b**(1/4) * np.exp(1j * (phi_3b + 2*np.pi*k)/4) for k in range(4)]
print("\n2.3 b) Корни:")
for i, z in enumerate(roots_3b): print(f"z_{i} = {z:.4f}")
plot_roots(roots_3b, "2.3 b)")

# c) sqrt[3]{ (1-5i)/(1+i) - 5/2 * (1+2i)/(-i) + 11 }
# Вычисляем внутреннее выражение
inner_3c = (1 - 5j)/(1 + 1j) - (5/2) * (1 + 2j)/(-1j) + 11
r_3c = abs(inner_3c)
phi_3c = np.angle(inner_3c)
roots_3c = [r_3c**(1/3) * np.exp(1j * (phi_3c + 2*np.pi*k)/3) for k in range(3)]
print("\n2.3 c) Корни:")
for i, z in enumerate(roots_3c): print(f"z_{i} = {z:.4f}")
plot_roots(roots_3c, "2.3 c)")

# d) sqrt[4]{1-i}
val_3d = 1 - 1j
r_3d = abs(val_3d)
phi_3d = np.angle(val_3d)
roots_3d = [r_3d**(1/4) * np.exp(1j * (phi_3d + 2*np.pi*k)/4) for k in range(4)]
print("\n2.3 d) Корни:")
for i, z in enumerate(roots_3d): print(f"z_{i} = {z:.4f}")
plot_roots(roots_3d, "2.3 d)")


```

```python
# 2.4. Изобразить множества точек на комплексной плоскости
print("Задание 2.4: Множества точек")

# Создаем сетку для графиков
x = np.linspace(-5, 5, 400)
y = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

# a) |z - i| + |z + i| < 4 (Эллипс с фокусами в (0,1) и (0,-1))
cond_a = np.abs(Z - 1j) + np.abs(Z + 1j) < 4
axes[0].contourf(X, Y, cond_a, levels=[0.5, 1.5], colors=['lightblue'])
axes[0].set_title("a) |z-i| + |z+i| < 4")

# b) Re(z(1-i)) < sqrt(2)
# z(1-i) = (x+iy)(1-i) = x - ix + iy + y = (x+y) + i(y-x)
# Re = x + y < sqrt(2) -> y < -x + sqrt(2)
cond_b = (X + Y) < np.sqrt(2)
axes[1].contourf(X, Y, cond_b, levels=[0.5, 1.5], colors=['lightgreen'])
axes[1].set_title("b) Re(z(1-i)) < sqrt(2)")

# c) |z+1| = |z+i| = |z-i|
# Уравнение |z+1| = |z+i| -> геометрическое место точек, равноудаленных от (-1,0) и (0,-1) -> прямая
# |z+1| = |z-i| -> равноудаленных от (-1,0) и (0,1) -> прямая
# Пересечение двух прямых даст точку.
# Возьмем контурные линии, где значения близки к 0.
Z1 = np.abs(Z + 1)
Z2 = np.abs(Z + 1j)
Z3 = np.abs(Z - 1j)
axes[2].contour(X, Y, Z1 - Z2, levels=[0], colors='red')
axes[2].contour(X, Y, Z1 - Z3, levels=[0], colors='blue')
axes[2].set_title("c) |z+1|=|z+i|=|z-i|")

# d) 0 < arg((i-z)/(z+i)) < pi/2
# Вычислим аргумент
W = (1j - Z) / (Z + 1j)
arg_W = np.angle(W)
# Верхняя граница 0 < arg < pi/2
cond_d = (arg_W > 0) & (arg_W < np.pi/2)
axes[3].contourf(X, Y, cond_d, levels=[0.5, 1.5], colors=['orange'])
axes[3].set_title("d) 0 < arg((i-z)/(z+i)) < pi/2")

# e) -2pi/3 <= arg((2-i)/(z+3i)) <= -5pi/6
W_e = (2 - 1j) / (Z + 3j)
arg_e = np.angle(W_e)
cond_e = (arg_e >= -2*np.pi/3) & (arg_e <= -5*np.pi/6)
axes[4].contourf(X, Y, cond_e, levels=[0.5, 1.5], colors=['purple'])
axes[4].set_title("e) arg((2-i)/(z+3i))")

# f) -2pi/3 <= arg((3-z)/(z+i)) <= -pi/3
W_f = (3 - Z) / (Z + 1j)
arg_f = np.angle(W_f)
cond_f = (arg_f >= -2*np.pi/3) & (arg_f <= -np.pi/3)
axes[5].contourf(X, Y, cond_f, levels=[0.5, 1.5], colors=['pink'])
axes[5].set_title("f) arg((3-z)/(z+i))")

for ax in axes:
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_aspect('equal')

plt.tight_layout()
plt.show()


```

```python
# 2.7. Обобщение биномиальных коэффициентов
print("Задание 2.7")

from scipy.special import gamma

def C_w_z(z, w):
    # C_w^z = Gamma(z+1) / (Gamma(w+1)*Gamma(z-w+1))
    return gamma(z + 1) / (gamma(w + 1) * gamma(z - w + 1))

# a) Найдите C_{1-i}^{1+i}
z_val = 1 + 1j
w_val = 1 - 1j
res_2_7a = C_w_z(z_val, w_val)
print(f"a) C_{w_val}^{z_val} = {res_2_7a}")

# b) Найдите Arg(C_{1-i}^{1+i})
arg_2_7b = np.angle(res_2_7a)
print(f"b) Arg = {arg_2_7b} рад ({np.degrees(arg_2_7b):.4f} град)")

# c) Построить линии уровня Arg(C_12 = const)
# Для построения графика нам нужно вычислить функцию на сетке
x_val = np.linspace(-5, 5, 400)
y_val = np.linspace(-5, 5, 400)
X_val, Y_val = np.meshgrid(x_val, y_val)
Z_val = X_val + 1j*Y_val

# Вычисляем C_1^z = Gamma(z+1) / Gamma(z)
# Упростим: Gamma(z+1)/Gamma(z) = z
# Так как дано C_1^z (w=1), то C_1^z = z
# Если в задании C_12, возможно имеется в виду C_{1}^2 или C_{1}^z?
# Предположим, что C_w^z = C_1^z = z
# Тогда Arg(z) = const - это лучи из начала координат.
# Судя по картинке в задании, там сложная фрактальная структура,
# значит там вычисляется не просто C_1^z, а что-то вроде C_z^{12} или C_w^z с фиксированным w=12?
# Перечитаем: "Построить линии уровня Arg(C_12 = const)".
# Скорее всего имеется в виду Arg( C_{12}^z ) или Arg( C_{z}^{12} ).
# Попробуем построить Arg(C_{1}^z) = Arg(z) для начала, но на картинке не лучи.
# На картинке из задания фрактал. Значит, скорее всего, вычисляется C_w^z где z=x+iy, а w=12?
# Или наоборот. Попробуем взять C_z^12 = Gamma(13) / (Gamma(z+1)*Gamma(13-z)).
# Попробуем нарисовать для C_w^z где z = x+iy, w = 0.5 (или w=12).
# В задании написано C_12, возможно это C_{12}^z? Но w - нижний индекс.

# Давайте построим Arg(C_{z}^{1.5}) в качестве примера фрактала.
# Или Arg(C_{z}^{z})?
# Реализуем безопасное вычисление гамма-функции для комплексных чисел.
def safe_gamma(z):
    try:
        return gamma(z)
    except:
        return np.nan + 1j*np.nan

def C_frac(z, w):
    # z, w - комплексные числа
    res = np.zeros_like(z, dtype=complex)
    for i in range(z.shape[0]):
        for j in range(z.shape[1]):
            zv = z[i,j]
            wv = w
            try:
                num = gamma(zv + 1)
                den = gamma(wv + 1) * gamma(zv - wv + 1)
                res[i,j] = num / den
            except:
                res[i,j] = np.nan + 1j*np.nan
    return res

# Зададим w = 1.5 (или 12, как в задании) - чтобы получить фрактал
w_const = 1.5
# Ограничим область, чтобы не улететь в бесконечность
C_val = C_frac(Z_val, w_const)
Arg_val = np.angle(C_val)

plt.figure(figsize=(8, 8))
plt.contourf(X_val, Y_val, Arg_val, levels=20, cmap='hsv')
plt.title(f"Линии уровня Arg(C_{w_const}^z)")
plt.colorbar(label='Arg')
plt.show()


```

```python
# 2.8. Матрица и корень из 1 степени 14
print("Задание 2.8")

# lambda - комплексный корень 14-й степени из 1
# lambda^14 = 1
# Мнимая часть > 0, действительная часть максимальна
# Корни 14-й степени: e^(2*pi*i*k/14), k=0..13
roots_14 = [np.exp(2j*np.pi*k/14) for k in range(14)]

# Отфильтруем те, у которых Im > 0
roots_pos_im = [r for r in roots_14 if r.imag > 1e-9]
# Найдем с максимальной действительной частью
lambda_val = max(roots_pos_im, key=lambda r: r.real)
print(f"λ = {lambda_val}")

# Матрицы
A = np.array([[7, 5], [6, 2]])
I = np.array([[1, 0], [0, 1]])

# Матрица A + lambda*I
M = A + lambda_val * I
print(f"Матрица A + λI:\n{M}")

# Определитель
det_M = np.linalg.det(M)
print(f"Определитель: {det_M}")
print(f"1) Действительная часть: {det_M.real:.5f}")
print(f"2) Мнимая часть: {det_M.imag:.5f}")


```

```python
# 2.9. Множества Жюлиа
print("Задание 2.9: Множества Жюлиа")

def julia_set(c, width=500, height=500, xmin=-2, xmax=2, ymin=-2, ymax=2, max_iter=100):
    julia = np.zeros((width, height))
    for ix in range(width):
        for iy in range(height):
            z = complex(ix / width * (xmax - xmin) + xmin,
                        iy / height * (ymax - ymin) + ymin)
            nit = 0
            while abs(z) <= 10 and nit < max_iter:
                z = z**2 + c
                nit += 1
            julia[ix, iy] = nit / max_iter
    return julia

# c = -0.1 + 0.65i
c1 = complex(-0.1, 0.65)
julia1 = julia_set(c1)

# c = 0.28 + 0.009i
c2 = complex(0.28, 0.009)
julia2 = julia_set(c2)

# Визуализация
fig, axes = plt.subplots(1, 2, figsize=(15, 7))
axes[0].imshow(julia1, cmap='flag', extent=[-2, 2, -2, 2], origin='lower')
axes[0].set_title(f"c = {c1}")
axes[0].axis('off')

axes[1].imshow(julia2, cmap='twilight_shifted', extent=[-2, 2, -2, 2], origin='lower')
axes[1].set_title(f"c = {c2}")
axes[1].axis('off')
plt.show()


# 2.9 (продолжение). Множества Жюлиа для других функций
print("\nЗадание 2.9 (продолжение): Другие функции")

def julia_custom(func, width=500, height=500, xmin=-2, xmax=2, ymin=-2, ymax=2, max_iter=100):
    julia = np.zeros((width, height))
    for ix in range(width):
        for iy in range(height):
            z = complex(ix / width * (xmax - xmin) + xmin,
                        iy / height * (ymax - ymin) + ymin)
            nit = 0
            while abs(z) <= 10 and nit < max_iter:
                try:
                    z = func(z)
                except ZeroDivisionError:
                    z = 10 # Вылетаем за границу
                nit += 1
            julia[ix, iy] = nit / max_iter
    return julia

# f(z) = z^5 - i (здесь i в степени 5? В задании написано f(z) = i=5-1 z^5-i, скорее всего z^5 - i)
# Судя по картинке, это z^5 - i
func2 = lambda z: z**5 - 1j
julia_func2 = julia_custom(func2)

# f(z) = z^2 + i / (1 - z^2) (в задании f(z) = i=2+ i 1-z^2, скорее всего z^2 + i/(1-z^2))
# Судя по картинке, это z^2 + i/(1-z^2)
def func3(z):
    if abs(1 - z**2) < 1e-10:
        return 100 # Отталкиваем
    return z**2 + 1j/(1 - z**2)

julia_func3 = julia_custom(func3)

fig, axes = plt.subplots(1, 2, figsize=(15, 7))
axes[0].imshow(julia_func2, cmap='gist_heat', extent=[-2, 2, -2, 2], origin='lower')
axes[0].set_title("f(z) = z^5 - i")
axes[0].axis('off')

axes[1].imshow(julia_func3, cmap='gist_stern', extent=[-2, 2, -2, 2], origin='lower')
axes[1].set_title("f(z) = z^2 + i/(1-z^2)")
axes[1].axis('off')
plt.show()


```

```python
# 2.10. Сфера Римана
print("Задание 2.10: Сфера Римана")

# Функция стереографической проекции
def stereographic_projection(x, y):
    # x, y - координаты на комплексной плоскости
    # Координаты на сфере (X, Y, Z)
    denom = x**2 + y**2 + 1
    X_sphere = 2 * x / denom
    Y_sphere = 2 * y / denom
    Z_sphere = (x**2 + y**2 - 1) / denom
    return X_sphere, Y_sphere, Z_sphere

# Создаем фигуру
fig = plt.figure(figsize=(12, 12))
ax = fig.add_subplot(111, projection='3d')

# Рисуем саму сферу
u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
X_sph = np.outer(np.cos(u), np.sin(v))
Y_sph = np.outer(np.sin(u), np.sin(v))
Z_sph = np.outer(np.ones(np.size(u)), np.cos(v))
ax.plot_surface(X_sph, Y_sph, Z_sph, color='y', alpha=0.3, edgecolor='k', linewidth=0.1)

# a) Прямая на плоскости (например, y = x)
t = np.linspace(-10, 10, 200)
x_line = t
y_line = t
X_line, Y_line, Z_line = stereographic_projection(x_line, y_line)
ax.plot(X_line, Y_line, Z_line, color='red', linewidth=2, label='Прямая')

# b) Окружность на плоскости (например, x^2 + y^2 = 4)
theta = np.linspace(0, 2*np.pi, 200)
x_circ = 2 * np.cos(theta)
y_circ = 2 * np.sin(theta)
X_circ, Y_circ, Z_circ = stereographic_projection(x_circ, y_circ)
ax.plot(X_circ, Y_circ, Z_circ, color='blue', linewidth=2, label='Окружность')

# c) Парабола на плоскости (например, y = x^2)
x_par = np.linspace(-2, 2, 200)
y_par = x_par**2
X_par, Y_par, Z_par = stereographic_projection(x_par, y_par)
ax.plot(X_par, Y_par, Z_par, color='green', linewidth=2, label='Парабола')

# d) Гипербола на плоскости (например, x^2 - y^2 = 1)
t_hyp = np.linspace(-2, 2, 200)
x_hyp = np.cosh(t_hyp)
y_hyp = np.sinh(t_hyp)
X_hyp, Y_hyp, Z_hyp = stereographic_projection(x_hyp, y_hyp)
ax.plot(X_hyp, Y_hyp, Z_hyp, color='purple', linewidth=2, label='Гипербола (ветвь 1)')
X_hyp2, Y_hyp2, Z_hyp2 = stereographic_projection(-x_hyp, y_hyp)
ax.plot(X_hyp2, Y_hyp2, Z_hyp2, color='purple', linewidth=2, linestyle='--', label='Гипербола (ветвь 2)')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Сфера Римана: кривые')
ax.legend()
plt.show()

# 2.10. Логарифмическая спираль на сфере Римана
# r = a * e^(b*phi)
# Возьмем a = 1, b = 0.2
phi_spiral = np.linspace(-5*np.pi, 5*np.pi, 500)
r_spiral = np.exp(0.2 * phi_spiral)
x_spiral = r_spiral * np.cos(phi_spiral)
y_spiral = r_spiral * np.sin(phi_spiral)

X_sp, Y_sp, Z_sp = stereographic_projection(x_spiral, y_spiral)

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X_sph, Y_sph, Z_sph, color='y', alpha=0.3, edgecolor='k', linewidth=0.1)
ax.plot(X_sp, Y_sp, Z_sp, color='red', linewidth=2)
ax.set_title('Логарифмическая спираль на сфере Римана')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```
